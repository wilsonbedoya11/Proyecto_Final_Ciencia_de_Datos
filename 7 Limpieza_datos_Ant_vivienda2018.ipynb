#### **3.2 Categorizar**

```python
import missingno as msno
msno.matrix(df, figsize=(8, 5), fontsize=10, color=(0.5, 0.5, 1))
```

```python
# Definir las condiciones  TIPO VIVIENDA
conditions = [
    (df['V_TIPO_VIV'] == 1),
    (df['V_TIPO_VIV'] == 2),
    (df['V_TIPO_VIV'] == 3),
    (df['V_TIPO_VIV'] == 4) | (df['V_TIPO_VIV'] == 5),
    (df['V_TIPO_VIV'] == 6) | (df['V_TIPO_VIV'] == 0) | (df['V_TIPO_VIV'].isna())
]

# Definir los valores que tomará la nueva columna
values = ['1 Casa', '2 Apartamento', '3 Tipo cuarto', '4 Vivienda tradicional Indigena 5 Vivienda tradicional Etnica (Afrocolombiana, Isleña)', '6 Otro (contenedor, carpa, embarcacion, vagon, cueva, refugio natural, etc.)' ]

# Crear la nueva columna usando numpy.select()
df['TIPO_VIVIENDA_CATEGORIA'] = np.select(conditions, values, default='6 Otro (contenedor, carpa, embarcacion, vagon, cueva, refugio natural, etc.)')
```

```python
# Definir las condiciones  TIPO PAREDES
conditions = [
    (df['V_MAT_PARED'] == 1) | (df['V_MAT_PARED'] == 2),
    (df['V_MAT_PARED'] == 3),
    (df['V_MAT_PARED'] == 4) | (df['V_MAT_PARED'] == 7),
    (df['V_MAT_PARED'] == 5),
    (df['V_MAT_PARED'] == 6),
    (df['V_MAT_PARED'] == 8),
    (df['V_MAT_PARED'] == 9),
    (df['V_MAT_PARED'] == 0) |(df['V_MAT_PARED'].isna())
]

# Definir los valores que tomará la nueva columna
values = ['1 Bloque, ladrillo, piedra, madera pulida 2 Concreto vaciado', '3 Material prefabricado', '4 Guadua 7 Caña, esterilla, otros vegetales',
          '5 Tapia pisada, bahareque, adobe', '6 Madera burda, tabla, tablon',
          '8 Materiales de deshecho (Zinc, tela, carton, latas, plasticos, otros)', '9 No tiene paredes', '99 No Informa']

# Crear la nueva columna usando numpy.select()
df['TIPO_PAREDES_CATEGORIA'] = np.select(conditions, values, default='9 No Informa')
```

```python
# Definir las condiciones  TIPO ENERGIA
conditions = [
    (df['VA_EE'] == 1),
    (df['VA_EE'] == 2),
    (df['VA_EE'] == 0) | (df['VA_EE'].isna())
]

# Definir los valores que tomará la nueva columna
values = ['1 SI', '2 NO', '99 No Informa']

# Crear la nueva columna usando numpy.select()
df['TIPO_ENERGIA_CATEGORIA'] = np.select(conditions, values, default='9 No Informa')
```

```python
# USO DEL JOIN PARA UNIR LA INFORMACION DEPARTAMENTO
df_merged = df.merge(df_departamentos,  left_on='U_DPTO', right_on='PK_DEPARTAMENTO', how='left')
```
