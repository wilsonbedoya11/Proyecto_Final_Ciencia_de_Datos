#### **3.2 Categorizar**

```python
import missingno as msno
msno.matrix(df, figsize=(8, 5), fontsize=10, color=(0.5, 0.5, 1))
```

```python
# Definir las condiciones  sexo
conditions = [
    (df['P_SEXO'] == 1),
    (df['P_SEXO'] == 2)
]

# Definir los valores que tomará la nueva columna
values = ['1 Hombre', '2 Mujer']

# Crear la nueva columna usando numpy.select()
df['SEXO_CATEGORIA'] = np.select(conditions, values, default='Desconocido')
```

```python
# Definir las condiciones  edad
conditions = [
    (df['P_EDADR'] == 1),
    (df['P_EDADR'] == 2),
    (df['P_EDADR'] == 3),
    (df['P_EDADR'] == 4),
    (df['P_EDADR'] == 5),
    (df['P_EDADR'] == 6),
    (df['P_EDADR'] == 7),
    (df['P_EDADR'] == 8),
    (df['P_EDADR'] == 9),
    (df['P_EDADR'] == 10),
    (df['P_EDADR'] == 11),
    (df['P_EDADR'] == 12),
    (df['P_EDADR'] == 13),
    (df['P_EDADR'] == 14),
    (df['P_EDADR'] == 15),
    (df['P_EDADR'] == 16),
    (df['P_EDADR'] == 17),
    (df['P_EDADR'] == 18),
    (df['P_EDADR'] == 19),
    (df['P_EDADR'] == 20),
    (df['P_EDADR'] == 21)
]

# Definir los valores que tomará la nueva columna
values = ['1 de 00 A 04 Años', '2 de 05 A 09 Años', '3 de 10 A 14 Años', '4 de 15 A 19 Años', '5 de 20 A 24 Años', '6 de 25 A 29 Años', '7 de 30 A 34 Años', '8 de 35 A 39 Años',
          '9 de 40 A 44 Años', '10 de 45 A 49 Años', '11 de 50 A 54 Años', '12 de 55 A 59 Años', '13 de 60 A 64 Años', '14 de 65 A 69 Años', '15 de 70 A 74 Años', '16 de 75 A 79 Años',
          '17 de 80 A 84 Años', '18 de 85 A 89 Años', '19 de 90 A 94 Años', '20 de 95 A 99 Años', '21 de 100 y más Años' ]

# Crear la nueva columna usando numpy.select()
df['EDAD_RANGO_CATEGORIA'] = np.select(conditions, values, default='Desconocido')
```

```python
# Definir las condiciones  alfabeta
conditions = [
    (df['P_ALFABETA'] == 1),
    (df['P_ALFABETA'] == 2),
    (df['P_ALFABETA'] == 9) | (df['P_ALFABETA'].isna())
]

# Definir los valores que tomará la nueva columna
values = ['1 SI', '2 NO', '9 No Informa']

# Crear la nueva columna usando numpy.select()
df['ALFABETA_CATEGORIA'] = np.select(conditions, values, default='9 No Informa')
```

```python
# Definir las condiciones  escolaridad
conditions = [
    (df['P_NIVEL_ANOSR'] == 1),
    (df['P_NIVEL_ANOSR'] == 2),
    (df['P_NIVEL_ANOSR'] == 3),
    (df['P_NIVEL_ANOSR'] == 4),
    (df['P_NIVEL_ANOSR'] == 5),
    (df['P_NIVEL_ANOSR'] == 6),
    (df['P_NIVEL_ANOSR'] == 7),
    (df['P_NIVEL_ANOSR'] == 8),
    (df['P_NIVEL_ANOSR'] == 9),
    (df['P_NIVEL_ANOSR'] == 10) | (df['P_NIVEL_ANOSR'] == 99) | (df['P_NIVEL_ANOSR'].isna())
]

# Definir los valores que tomará la nueva columna
values = ['1 Preescolar', '2 Básica primaria', '3 Básica secundaria', '4 Media academica o clasica', '5 Media tecnica',
          '6 Normalista', '7 Técnica profesional o Tecnológica', '8 Universitario', '9 Especialización, maestría, doctorado',
          '99 No Informa']

# Crear la nueva columna usando numpy.select()
df['EDUCACION_NIVEL_EDUCATIVO_CATEGORIA'] = np.select(conditions, values, default='99 No Informa')
```
