## **1. VIVIENDA**

```python
# IMPORTAR ARCHIVOS PERSONAS 2018
df1 = pd.read_csv('/content/VIVIENDA_2005_CAT.csv')
df2 = pd.read_csv('/content/VIVIENDA_2018_CAT.csv')
```

```python
# ELIMINIAR CAMPOS NO COMUNES
df1 = df1.drop(columns=['LLAVE_VIV_N'])
df2 = df2.drop(columns=['COD_ENCUESTAS'])
```
