#### **3.1 Identificación de columnas y eliminación**

```python
# PERSONAS 2005

df = df.drop(columns=['RECORD'])
df = df.drop(columns=['I0B1_LOCALIDAD'])
df = df.drop(columns=['SEC_AGAD'])
df = df.drop(columns=['I0F_EDIFIC'])
df = df.drop(columns=['I0G_UCENSAL'])
df = df.drop(columns=['VC01B_TOT_HOG'])
df = df.drop(columns=['V04A_ELI_BASURA'])
df = df.drop(columns=['V05B2_ALCANTARI'])
df = df.drop(columns=['V05B3_ACUEDUCTO'])
df = df.drop(columns=['V05B4_GAS_NAT'])
df = df.drop(columns=['V05B5_TELEFONO'])
df = df.drop(columns=['V06A_TIPO_SERSA'])
df = df.drop(columns=['V07A_AGUA_SERVI'])
df = df.drop(columns=['V08A_NRO_BANOS'])
df = df.drop(columns=['V09A_EXISTE_COCINA'])
df = df.drop(columns=['FACT_EXP_CAL_V_N'])
df = df.drop(columns=['I0C_CLASER'])
df = df.drop(columns=['V03A_MAT_PISOS'])
```

#### **3.2 Categorizar**

```python
import missingno as msno
msno.matrix(df, figsize=(8, 5), fontsize=10, color=(0.5, 0.5, 1))
```

```python
# Definir las condiciones  TIPO VIVIENDA
conditions = [
    (df['V01B_TIP_VIV'] == 1),
    (df['V01B_TIP_VIV'] == 2),
    (df['V01B_TIP_VIV'] == 3),
    (df['V01B_TIP_VIV'] == 4),
    (df['V01B_TIP_VIV'] == 5)
]

# Definir los valores que tomará la nueva columna
values = ['1 Casa', '2 Casa indigena', '3 Apartamento', '4 Tipo cuarto',  '5 Otro tipo de vivienda' ]

# Crear la nueva columna usando numpy.select()
df['TIPO_VIVIENDA_CATEGORIA'] = np.select(conditions, values, default='Desconocido')
```

```python
# Definir las condiciones  TIPO PAREDES
conditions = [
    (df['V02A_MAT_PARED'] == 1),
    (df['V02A_MAT_PARED'] == 2),
    (df['V02A_MAT_PARED'] == 3),
    (df['V02A_MAT_PARED'] == 4),
    (df['V02A_MAT_PARED'] == 5),
    (df['V02A_MAT_PARED'] == 6),
    (df['V02A_MAT_PARED'] == 7),
    (df['V02A_MAT_PARED'] == 9) | (df['V02A_MAT_PARED'] == 0) |(df['V02A_MAT_PARED'].isna())
]

# Definir los valores que tomará la nueva columna
values = ['1 Bloque, ladrillo, piedra, madera pulida', '2 Tapia pisada, adobe, bahareque', '3 Madera burda, tabla, tablón', '4 Material prefabricado',  '5 Guadua, caña, esterilla, otros vegetales',
          '6 Zinc, tela, cartón, latas, desechos, plásticos', '7 Sin paredes', '9 No Informa']

# Crear la nueva columna usando numpy.select()
df['TIPO_PAREDES_CATEGORIA'] = np.select(conditions, values, default='9 No Informa')
```

```python
# Definir las condiciones  TIPO ENERGIA
conditions = [
    (df['V05B1_ELECTRICA'] == 1),
    (df['V05B1_ELECTRICA'] == 2)
]

# Definir los valores que tomará la nueva columna
values = ['1 SI', '2 NO']

# Crear la nueva columna usando numpy.select()
df['TIPO_ENERGIA_CATEGORIA'] = np.select(conditions, values, default='9 No Informa')
```

```python
# ARCHIVO CON LA INFORMACION DE LOS DEPARTAMENTOS
# Información obtenida de:

# https://www.datos.gov.co/Mapas-Nacionales/DEPARTAMENTOS/25e3-npng/about_data
# https://geoportal.dane.gov.co/descargas/divipola/DIVIPOLA_Municipios.xlsx

df_departamentos = pd.read_csv("/content/Departamentos.csv", delimiter=';')
df_departamentos.head()
```

```python
# ARCHIVO CON LA INFORMACION DE LOS MUNICIPIOS
df_municipios = pd.read_csv("/content/MunicipiosAntioquia1.csv", delimiter=',')
df_municipios.head()
```

```python
# USO DEL JOIN PARA UNIR LA INFORMACION DEPARTAMENTO
df_merged = df.merge(df_departamentos,  left_on='I0A_DPTO', right_on='PK_DEPARTAMENTO', how='left')
```

```python
df_merged.head()
```

```python
# USO DEL JOIN PARA UNIR LA INFORMACION MUNICIPIOS
df_merged = df_merged.merge(df_municipios,  left_on='I0B_MPIO', right_on='PK_MUNICIPIO_SIN_DEPARTAMENTO', how='left')
```

```python
df_merged.head()
```

```python
df_final= df_merged.copy()
```

```python
df_final.columns
```

```python
df_final = df_final.drop(columns=['I0A_DPTO'])
df_final = df_final.drop(columns=['I0B_MPIO'])
df_final = df_final.drop(columns=['V01B_TIP_VIV'])
df_final = df_final.drop(columns=['V02A_MAT_PARED'])
df_final = df_final.drop(columns=['V05B1_ELECTRICA'])
df_final = df_final.drop(columns=['PK_DEPARTAMENTO'])
df_final = df_final.drop(columns=['FK_REGION'])
df_final = df_final.drop(columns=['PK_MUNICIPIO'])
df_final = df_final.drop(columns=['PK_MUNICIPIO_SIN_DEPARTAMENTO'])
df_final = df_final.drop(columns=['LONGITUD'])
df_final = df_final.drop(columns=['LACTITUD'])
df_final = df_final.drop(columns=['FK_DEPARTAMENTO'])
```
