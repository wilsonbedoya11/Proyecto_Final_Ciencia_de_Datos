## **UNION DE ARCHIVOS PERSONAS  2018**

```python
import pandas as pd
```

```python
# IMPORTAR ARCHIVOS PERSONAS 2018
df1 = pd.read_csv('/content/CensoPersonas2018_CategoriaAnt.csv')
df2 = pd.read_csv('/content/CensoPersonas2018_CategoriaCun.csv')
```

```python
df1.shape
```

```python
df2.shape
```

```python
df1.shape[0] + df2.shape[0]
```

```python
# Concatenar los DataFrame (Unión vertical)
df_resultado = pd.concat([df1, df2], ignore_index=True)

df_resultado.shape
```
